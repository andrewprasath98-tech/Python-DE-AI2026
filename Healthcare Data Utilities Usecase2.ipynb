##Task 1: Define Base Paths Using Python Variables

```python

# Define the base path
base_path = "/Volumes/workspace/default/volumewd36"

# Define directory paths
raw_path = f"{base_path}/raw"
processed_path = f"{base_path}/processed"
archive_path = f"{base_path}/archive"

# Display the paths
print("Base Path:", base_path)
print("Raw Path:", raw_path)
print("Processed Path:", processed_path)
print("Archive Path:", archive_path)
```

##Task 2: Use dbutils to Create Directories

```python

# Create directories
dbutils.fs.mkdirs(raw_path)
dbutils.fs.mkdirs(processed_path)
dbutils.fs.mkdirs(archive_path)

print("Directories created successfully!")
```

##Task 3: dbutils Use Case – Create Sample Healthcare File

```python

sample_data = """patient_id,patient_name,age,gender
1,John Doe,68,M
2,Jane Smith,54,F
"""
```

```python

# Define the file path
file_path = f"{raw_path}/healthcare_data.csv"

# Write sample data into the file
dbutils.fs.put(file_path, sample_data, overwrite=True)

print("Healthcare CSV file created successfully!")
```

```python

display(dbutils.fs.ls(raw_path))
```

```python

print(dbutils.fs.head(file_path))
```

##Task 4: dbutils Use Case – List the Files Created

##Task 5: dbutils Use Case – Copy File (Raw → Processed)

```python
source_path = f"{raw_path}/healthcare_data.csv"

destination_path = f"{processed_path}/healthcare_data.csv"

# Copy file from raw to processed
dbutils.fs.cp(source_path, destination_path)

print("File copied successfully!")
display(dbutils.fs.ls(processed_path))
print(dbutils.fs.head(destination_path))
```

# Task 6: Databricks dbutils.widgets Use Case – Create Dropdown and Text Widgets

```python
# Create environment dropdown widget
dbutils.widgets.dropdown(
    "environment",
    "dev",
    ["dev", "qa", "prod"],
    "Select Environment"
)
```

```python
# Create owner name text widget
dbutils.widgets.text(
    "owner",
    "Andrew",
    "Enter Owner Name"
)
```

```python
# Get widget values
environment = dbutils.widgets.get("environment")
owner = dbutils.widgets.get("owner")

print("Environment:", environment)
print("Owner Name:", owner)
```

# dbutils widget Usecase – Read Widget Values environment and owner and print in the screen

```python

# Read environment widget value
environment = dbutils.widgets.get("environment")

# Read owner widget value
owner = dbutils.widgets.get("owner")

# Print widget values
print("Environment:", environment)
print("Owner Name:", owner)
```

# Task 8: dbutils Use Case – Move the Processed File to Archive

```python
# Source file path
source_path = f"{processed_path}/healthcare_data.csv"

# Destination file path
destination_path = f"{archive_path}/healthcare_data.csv"

# Move file from processed to archive
dbutils.fs.mv(source_path, destination_path)

print("File moved successfully from processed to archive!")
display(dbutils.fs.ls(archive_path))
display(dbutils.fs.ls(processed_path))
```
