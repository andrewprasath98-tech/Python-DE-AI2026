#Task 6: Create a PySpark DataFrame (df1) from the CSV File

```python

%python

# Define the CSV file path
file_path = "/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv"

# Read CSV file into PySpark DataFrame
df1 = spark.read \
    .format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(file_path)

# Display the DataFrame
display(df1)
```

#Task 7: Write PySpark DataFrame into a Databricks Table
You need to save the DataFrame df1 into a Databricks table named default.mobile_os_usage.

```python

%python

# Delta table column names cannot contain spaces — rename them first
df1_clean = df1.withColumnRenamed("Mobile Operating System", "Mobile_Operating_System") \
    .withColumnRenamed("Percent of Usage", "Percent_of_Usage")

# Write DataFrame into Databricks table
df1_clean.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("default.mobile_os_usage")

print("Data successfully written into the table!")
```
