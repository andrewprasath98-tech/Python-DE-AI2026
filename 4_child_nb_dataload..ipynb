```python

%py

import requests

# API URL
url = "https://public.tableau.com/app/sample-data/mobile_os_usage.csv"

# Perform API call
response = requests.get(url)

# Validate API response
response.raise_for_status()

# Convert response into text variable
data = response.text

# Display first 500 characters
print(data[:500])
```

```python

%py

# Define volume path
volume_path = "/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv"

# Write CSV data into volume
dbutils.fs.put(volume_path, data, overwrite=True)

print("CSV file successfully loaded into the volume!")
```
